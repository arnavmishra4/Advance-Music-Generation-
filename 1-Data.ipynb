# Preparing the dataset 

For data preprationg , we are going to extract these information from the MIDI files:
- Pitch 
- velocity 
- Start Time 
- End time 
- Instrument 

```python
import os
import pretty_midi

# Function to convert MIDI to Event-Based Encoding
def midi_to_events(midi_file):
    try:
        pm = pretty_midi.PrettyMIDI(midi_file)
    except Exception as e:
        print(f"❌ Skipping {midi_file}: {e}")
        return None  # Skip invalid MIDI files
    
    events = []
    for instrument in pm.instruments:
        for note in instrument.notes:
            events.append({
                "pitch": int(note.pitch),  
                "velocity": int(note.velocity),  
                "start": float(note.start),  
                "end": float(note.end),  
                "instrument": int(instrument.program)  
            })
    
    events.sort(key=lambda x: x["start"])
    return events
def process_midi_dataset(root_dir):
    dataset = []

    for composer in os.listdir(root_dir):
        composer_path = os.path.join(root_dir, composer)

        if os.path.isdir(composer_path): 
            print(f"🎵 Processing {composer}...")  
            
            for midi_file in os.listdir(composer_path):
                if midi_file.endswith(".mid"):
                    file_path = os.path.join(composer_path, midi_file)

                    events = midi_to_events(file_path)
                    if events is None: 
                        continue  

                    dataset.append({
                        "composer": composer,
                        "events": events
                    })
    
    return dataset

root_dir = "D:\PersonalProjects\Music Generation\musicnet_midis\musicnet_midis" # Change to your actual directory
dataset = process_midi_dataset(root_dir)

if dataset:
    print(f"\n✅ Successfully processed {len(dataset)} MIDI files!")
    print("Sample:", dataset[0]) 
else:
    print("❌ No valid MIDI files found.")
```

```python
import pickle

with open('dataset.pkl', 'wb') as f:
    pickle.dump(dataset, f)
```
